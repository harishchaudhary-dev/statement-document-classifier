```python
# Notebook 02: Model Training & Evaluation
import pickle
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report

dataset = [
    ("Account Statement Balance Account Number Credits Debits", "bank_statement"),
    ("Credit Card Summary Minimum Amount Due Payment Due Date", "credit_card_statement"),
    ("Tax Return Form W-2 Gross Pay Tax Withheld", "tax_document"),
    ("INVOICE Invoice Number Bill To Total Amount", "invoice")
]

df = pd.DataFrame(dataset, columns=["text", "label"])
pipeline = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('clf', MultinomialNB())
])

pipeline.fit(df['text'], df['label'])

with open('../src/model.pkl', 'wb') as f:
    pickle.dump(pipeline, f)

print("Model trained and saved to ../src/model.pkl")
```
